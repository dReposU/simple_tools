# Set SSH from 0

```python
acc_email = "a@g.c"
acc_name = "foo"

# ssh_name
type = "ed25519"
service = "github"
device = "laptopFoo"

ssh_name = f"id_{type}_{service}_{acc_name}_{device}"

# Host

abbreviaton_name = "f"
abbreviaton_device = "LF"

host_alias = f"{service}_{abbreviaton_name}_{abbreviaton_device}"
host_name = "service.com"


# git@host:name/repo.git
repo_name = "foo"

git_url = f"git@{host_name}:{acc_name}/{repo_name}.git" # remote
```

```python
print(f"""
      CMD
      1.
      ssh-keygen -t ed25519 -C "{acc_email}"
      """)
```

```python
ssh_path = f"c/Users/.../.ssh/{ssh_name}"
```

```python
print(f"""
      CMD
      1.
      
      {ssh_path}
      
      2. 
      eval "$(ssh-agent -s)"
      ssh-add ~/.ssh/{ssh_name}
      
      3.
      cat ~/.ssh/{ssh_name}.pub
      
      GITBASH
      4.
      
      cd ~/.ssh
      nano config
      Host {host_alias}
        HostName {host_name}
        User git
        IdentityFile ~/.ssh/{ssh_name}
        
      ssh -T {host_alias}
      
      VSC (Local repo)
      5.
      git remote set-url origin git@{host_alias}:{acc_name}/{repo_name}.git
      git remote -v
      
      6.
      git config user.name "{acc_name}"
      git config user.email "{acc_email}"
      
      GIT BASH
      8. 
      8.1.
      touch ~/.ssh/allowed_signers
      git config --global gpg.ssh.allowedSignersFile ~/.ssh/allowed_signers
      """)
```

```python
# Allow signer <principal> <tipo_clave> <clave_base64> [comentario (device)]
print(f""" 
      Extract type and key from the cat command and add it to type_and_key variable. 
      Delete the email (last part of the output).
      
      cat {ssh_name}.pub
      """)
type_and_key = "ssh-ed25519 AAA..."

signer = f"{acc_email} {type_and_key} {device}"
```

```python
print(f"""
      GIT BASH
      8.3.
      
      echo "{signer}" >> ~/.ssh/allowed_signers
      
      git config commit.gpgsign true
      git config gpg.format ssh
      git config user.signingkey ~/.ssh/{ssh_name}.pub
      """)
```
